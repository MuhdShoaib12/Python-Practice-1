```python
n1 = int(input("Enter first number: "))
n2 = int(input("Enter second number: "))
op = input("Enter Arithmetic Operation Symbol (+,-,*,/)")

match op :
    case "+" :
        print(n1 + n2)
    case "-" :
        print(n1 - n2)
    case "*" :
        print(n1 * n2)
    case "/" :
        print(n1 / n2)            
    case _ :
        print("Invalid")
```

```python
day = int(input("Enter the number of day between(1-7): "))

match day :
    case 1:
        print("SUNDAY")
    case 2:
        print("MONDAY")
    case 3:
        print("TUESDAY")
    case 4:
        print("WENSDAY") 
    case 5:
        print("THURSHDAY")
    case 6:
        print("FRIDAY")
    case 7:
        print("SATURDAY")
    case _:
        print("Invalid")                                   
```

```python
light = input("Enter Traffic Light Color: (red/yellow/green): ").lower()
match light:
    case "red" :
        print("Just ignore and go!")
    case "yellow" :
        print("Go faster!")
    case "green" :
        print("Definitely go!") 
    case _:
        print("Invalid")           
```

```python
print("Menu\n1. Celsius to Fahrenheit\n2. Fahrenheit to Celsius")
option = int(input("Your choice: "))
if option not in [1,2]:
    print("Invalid choice....")
else:
    temp = float(input("Enter Temperature: "))
    match option:
        case 1:
            f = temp * (9/5) + 32
            print(f"{temp} C is {f} F. ")
        case 2:
            c = (temp - 32 ) * (5/9)
            print(f"{temp} F is {c} C.")        
```
